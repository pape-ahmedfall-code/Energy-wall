## 2. Fonctions du modèle

À implémenter : compute d'entraînement, compute d'inférence, énergie, CO2e, eau.

```python
def compute_entrainement_flops(n_parametres, n_tokens):
    """C_train ~ 6 * N * D (FLOPs)."""
    return 6 * n_parametres * n_tokens


def compute_inference_flops(n_parametres, n_tokens_servis):
    """C_inf ~ 2 * N par token servi (FLOPs)."""
    return 2 * n_parametres * n_tokens_servis


def energie_kwh(compute_flops, flops_par_joule, taux_utilisation, pue):
    """Énergie en kWh : compute / (efficacité * utilisation) * PUE, converti de joules en kWh."""
    joules = compute_flops / (flops_par_joule * taux_utilisation) * pue
    return joules / 3.6e6


def emissions_kgco2e(energie_kwh_, intensite_gco2_par_kwh):
    return energie_kwh_ * intensite_gco2_par_kwh / 1000


def eau_litres(energie_kwh_, wue_l_par_kwh):
    return energie_kwh_ * wue_l_par_kwh
```
